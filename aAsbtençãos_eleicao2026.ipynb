Baixando dados do TSE - ELEITORADO E LOCAIS DE VOTAÇÃO

```python
import requests
import zipfile
import io
import pandas as pd

# --------------------------------------------------
# DOWNLOADS
# --------------------------------------------------

URL_LOCAIS = (
    "https://cdn.tse.jus.br/estatistica/sead/odsele/"
    "eleitorado_locais_votacao/"
    "eleitorado_local_votacao_2026.zip"
)

URL_VOTOS = (
    "https://cdn.tse.jus.br/estatistica/sead/odsele/"
    "votacao_secao/"
    "votacao_secao_2026_BR.zip"
)

print("Baixando locais...")
resp_locais = requests.get(URL_LOCAIS, timeout=120)

print("Baixando votação...")
resp_votos = requests.get(URL_VOTOS, timeout=120)

print("Status locais:", resp_locais.status_code)
print("Status votos :", resp_votos.status_code)

zf_locais = zipfile.ZipFile(io.BytesIO(resp_locais.content))
zf_votos = zipfile.ZipFile(io.BytesIO(resp_votos.content))

print("\nArquivos locais:")
print(zf_locais.namelist())

print("\nArquivos votação:")
print(zf_votos.namelist())
```

Visualizar o eleitorado

```python
print("\nLendo eleitorado...")

with zf_locais.open(
    "eleitorado_local_votacao_2026_BRASIL.csv"
) as f:

    eleitorado = pd.read_csv(
        f,
        sep=";",
        encoding="latin1",
        usecols=[
            "SG_UF",
            "CD_MUNICIPIO",
            "NM_MUNICIPIO",
            "NR_ZONA",
            "NR_SECAO",
            "NR_LOCAL_VOTACAO",
            "NM_LOCAL_VOTACAO",
            "QT_ELEITOR_ELEICAO_FEDERAL"
        ]
    )

print(eleitorado.shape)

display(eleitorado.head())
```

Visualizar o local de votação

```python
print("\nLendo votação...")

with zf_votos.open(
    "votacao_secao_2026_BR.csv"
) as f:

    votos = pd.read_csv(
        f,
        sep=";",
        encoding="latin1",
        usecols=[
            "SG_UF",
            "CD_MUNICIPIO",
            "NR_ZONA",
            "NR_SECAO",
            "NR_LOCAL_VOTACAO",
            "NM_LOCAL_VOTACAO",
            "NM_VOTAVEL",
            "QT_VOTOS"
        ]
    )

print(votos.shape)

display(votos.head())
```

```python
print(
    votos["NM_VOTAVEL"]
    .drop_duplicates()
    .sort_values()
    .tolist()
)
```

Verificar quem compareceu para votar

```python
comparecimento = (

    votos

    .groupby(
        [
            "SG_UF",
            "CD_MUNICIPIO",
            "NR_ZONA",
            "NR_SECAO",
            "NR_LOCAL_VOTACAO",
            "NM_LOCAL_VOTACAO"
        ],
        as_index=False
    )

    ["QT_VOTOS"]

    .sum()

)

comparecimento.rename(
    columns={
        "QT_VOTOS":"COMPARECIMENTO"
    },
    inplace=True
)

print(comparecimento.shape)

display(comparecimento.head())
```

Calcular, então , a Abstenção

```python
abstencao = eleitorado.merge(

    comparecimento,

    on=[
        "SG_UF",
        "CD_MUNICIPIO",
        "NR_ZONA",
        "NR_SECAO",
        "NR_LOCAL_VOTACAO",
        "NM_LOCAL_VOTACAO"
    ],

    how="left"
)

abstencao["COMPARECIMENTO"] = (
    abstencao["COMPARECIMENTO"]
    .fillna(0)
)

abstencao["ABSTENCAO"] = (

    abstencao["QT_ELEITOR_ELEICAO_FEDERAL"]

    -

    abstencao["COMPARECIMENTO"]

)

abstencao["PCT_ABSTENCAO"] = (

    100

    *

    abstencao["ABSTENCAO"]

    /

    abstencao["QT_ELEITOR_ELEICAO_FEDERAL"]

).round(2)

print(abstencao.shape)

display(abstencao.head())
```

Juntar com dados da escola/local de votação - Por CIDADES


```python
abstencao_local = (

    abstencao

    .groupby(
        [
            "SG_UF",
            "CD_MUNICIPIO",
            "NM_MUNICIPIO",
            "NR_LOCAL_VOTACAO",
            "NM_LOCAL_VOTACAO"
        ],
        as_index=False
    )

    .agg(
        {
            "QT_ELEITOR_ELEICAO_FEDERAL":"sum",
            "COMPARECIMENTO":"sum",
            "ABSTENCAO":"sum"
        }
    )

)

abstencao_local["PCT_ABSTENCAO"] = (

    100

    *

    abstencao_local["ABSTENCAO"]

    /

    abstencao_local["QT_ELEITOR_ELEICAO_FEDERAL"]

).round(2)

display(
    abstencao_local
    .sort_values(
        "QT_ELEITOR_ELEICAO_FEDERAL",
        ascending=False
    )
    .head(20)
)
```

Coordenadas dos Locais de Votação - Facilitar para fazer os mapas

```python
# --------------------------------------------------
# COORDENADAS DOS LOCAIS DE VOTAÇÃO
# --------------------------------------------------

import pandas as pd

with zf_locais.open(
    "eleitorado_local_votacao_2026_BRASIL.csv"
) as f:

    coordenadas = pd.read_csv(
        f,
        sep=";",
        encoding="latin1",
        usecols=[
            "SG_UF",
            "CD_MUNICIPIO",
            "NM_MUNICIPIO",
            "NR_ZONA",
            "NR_LOCAL_VOTACAO",
            "NM_LOCAL_VOTACAO",
            "DS_ENDERECO",
            "NM_BAIRRO",
            "NR_LATITUDE",
            "NR_LONGITUDE"
        ]
    )

# elimina duplicidades de seção
coordenadas = coordenadas.drop_duplicates(
    subset=[
        "SG_UF",
        "CD_MUNICIPIO",
        "NR_ZONA",
        "NR_LOCAL_VOTACAO"
    ]
).copy()

# converte coordenadas
coordenadas["NR_LATITUDE"] = (
    coordenadas["NR_LATITUDE"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

coordenadas["NR_LONGITUDE"] = (
    coordenadas["NR_LONGITUDE"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

coordenadas["NR_LATITUDE"] = pd.to_numeric(
    coordenadas["NR_LATITUDE"],
    errors="coerce"
)

coordenadas["NR_LONGITUDE"] = pd.to_numeric(
    coordenadas["NR_LONGITUDE"],
    errors="coerce"
)

print("Locais únicos:", len(coordenadas))

print(
    "Locais com coordenadas:",
    coordenadas[
        ["NR_LATITUDE","NR_LONGITUDE"]
    ]
    .notna()
    .all(axis=1)
    .sum()
)

display(coordenadas.head(20))
```

Agragar votos por escola

```python
votos_escola = (

    votos

    .groupby(
        [
            "SG_UF",
            "CD_MUNICIPIO",
            "NR_LOCAL_VOTACAO",
            "NM_LOCAL_VOTACAO",
            "NM_VOTAVEL"
        ],
        as_index=False
    )

    ["QT_VOTOS"]

    .sum()

)

print(votos_escola.shape)

display(votos_escola.head())
```

```python
votos_wide = votos_escola.pivot_table(

    index=[
        "SG_UF",
        "CD_MUNICIPIO",
        "NR_LOCAL_VOTACAO",
        "NM_LOCAL_VOTACAO"
    ],

    columns="NM_VOTAVEL",

    values="QT_VOTOS",

    aggfunc="sum",

    fill_value=0

).reset_index()

print(votos_wide.shape)

display(votos_wide.head())
```

```python
mapa_final = coordenadas.merge(

    abstencao_local,

    on=[
        "SG_UF",
        "CD_MUNICIPIO",
        "NR_LOCAL_VOTACAO",
        "NM_LOCAL_VOTACAO"
    ],

    how="inner"

)

mapa_final = mapa_final.merge(

    votos_wide,

    on=[
        "SG_UF",
        "CD_MUNICIPIO",
        "NR_LOCAL_VOTACAO",
        "NM_LOCAL_VOTACAO"
    ],

    how="left"

)

print(mapa_final.shape)

display(mapa_final.head())
```

```python
col_lula = [c for c in mapa_final.columns if "LULA" in c.upper()][0]

col_bol = [c for c in mapa_final.columns if "BOLSONARO" in c.upper()][0]

print("Lula:", col_lula)
print("Bolsonaro:", col_bol)
```

```python
mapa_final["TOTAL_VOTOS"] = (

    mapa_final["COMPARECIMENTO"]

)

mapa_final["PCT_LULA"] = (

    100

    * mapa_final[col_lula]

    / mapa_final["COMPARECIMENTO"]

).round(2)

mapa_final["PCT_BOLSONARO"] = (

    100

    * mapa_final[col_bol]

    / mapa_final["COMPARECIMENTO"]

).round(2)

mapa_final["MARGEM"] = (

    mapa_final["PCT_LULA"]

    -

    mapa_final["PCT_BOLSONARO"]

).round(2)
```

```python
def inteiro(x):
    try:
        return f"{int(x):,}".replace(",", ".")
    except:
        return "0"
```

```python
import json

geojson = {
    "type": "FeatureCollection",
    "features": []
}

for _, r in mapa_final.iterrows():

    geojson["features"].append({

        "type": "Feature",

        "geometry": {
            "type": "Point",
            "coordinates": [
                float(r["NR_LONGITUDE"]),
                float(r["NR_LATITUDE"])
            ]
        },

        "properties": {

            "escola":
                str(r["NM_LOCAL_VOTACAO"]),

            "municipio":
                str(r["NM_MUNICIPIO"]),

            "bairro":
                str(r.get("NM_BAIRRO","")),

            "aptos":
                int(r["QT_ELEITOR_ELEICAO_FEDERAL"]),

            "comparecimento":
                int(r["COMPARECIMENTO"]),

            "abstencao":
                int(r["ABSTENCAO"]),

            "pct_abstencao":
                round(float(r["PCT_ABSTENCAO"]),2),

            "lula":
                int(r["LUIZ INÁCIO LULA DA SILVA"]),

            "bolsonaro":
                int(r["FLAVIO NANTES BOLSONARO"]),

            "branco":
                int(r["VOTO BRANCO"]),

            "nulo":
                int(r["VOTO NULO"])
        }
    })

with open(
    "/content/tse2026/escolas.geojson",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        geojson,
        f,
        ensure_ascii=False
    )

print(
    "GeoJSON salvo:"
)

print(
    "/content/tse2026/escolas.geojson"
)

print(
    "Total de escolas:",
    len(geojson["features"])
)
```
